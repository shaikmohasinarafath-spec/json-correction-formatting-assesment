###Question 1: Correcting a Malformed JSON String

The following string is intended to be valid JSON but contains several errors. Identify all errors, correct them, and parse the JSON into a Python dictionary. Print the resulting dictionary.

````python
malformed_json_q1 = """{  'name': 'Alice',  age: 30,  "city": "New York",  "hobbies": ["reading", "hiking",]} """
````

````python
import json
import re

# The original malformed JSON from the previous cell
# malformed_json_q1 = """{  'name': 'Alice',  age: 30,  "city": "New York",  "hobbies": ["reading", "hiking",]} """

# Step 1: Replace all single quotes with double quotes. This addresses both single-quoted keys and values.
corrected_json_step1 = re.sub(r"'", r'"', malformed_json_q1)

# Step 2: Enclose unquoted keys in double quotes
# This regex looks for word characters (keys) followed by a colon, not preceded by a double quote
corrected_json_step2 = re.sub(r'(?<!")([a-zA-Z0-9_]+)(?=\s*:)', r'"\1"', corrected_json_step1)

# Step 3: Remove trailing commas in arrays and objects
# This regex matches a comma followed by optional whitespace and then a ']' or '}'
corrected_json_final = re.sub(r',\s*([\]}])', r'\1', corrected_json_step2)

# Parse the corrected JSON string
data = json.loads(corrected_json_final)

# Print the resulting dictionary
print(data)
````

Explanation: What was wrong and how did you fix it?

The JSON had single quotes around the same, a missing quote around the age key, and a trailing comma in the hobbies list. I corrected these JSON syntax errors and used json.loads() to parse the corrected JSON into a Python dictionary.

#Question 2: Formatting a Python Dictionary to JSON

Convert the given Python dictionary to a JSON string with the following requirements:
- Indentation of 4 spaces
- Keys sorted alphabetically
- Non-ASCII characters preserved (ensure_ascii=False)
- Print the resulting JSON string.

````python
import json

data = {"name": "José", "age": 30, "city": "München", "hobbies": ["reading", "hiking"]}

json_string = json.dumps(data, indent=4, sort_keys=True, ensure_ascii=False)
print(json_string)
````

## Question 3: Extracting and Formatting Nested JSON
Given the following JSON string representing a list of users:
1.   Extract the email of the user with id=2 and print it.
2.   Pretty-print the entire JSON with indentation of 2 spaces.



````python
import json

json_str = '''
[
  {"id": 1, "name": "Alice", "email": "alice@example.com"},
  {"id": 2, "name": "Bob", "email": "bob@example.com"},
  {"id": 3, "name": "Charlie", "email": "charlie@example.com"}
]
'''

data = json.loads(json_str)

# Extract and print the email of the user with id=2
email = next(user['email'] for user in data if user['id'] == 2)
print(email)

# Pretty-print the entire JSON with indentation of 2 spaces
pretty_json = json.dumps(data, indent=2)
print(pretty_json)
````

## Question 4: Handling Duplicate Keys

The following JSON string contains duplicate keys:

```
{"a": 1, "b": 2, "a": 3}
```


1.   Explain what happens when you parse it with json.loads.
2.   Write code to parse it and merge duplicate keys by summing their values. Print the result.




````python
import json

dup_json = '{"a": 1, "b": 2, "a": 3}'

parsed = json.loads(dup_json)
print("Normal json.loads():", parsed)

# Corrected typo: object_paris_hook -> object_pairs_hook
pairs = json.loads(dup_json, object_pairs_hook=list)

# Merge duplicate keys by summing their values
merged = {}
for key, value in pairs:
    merged[key] = merged.get(key, 0) + value

print("Merged result:", merged)
````

Explanation: What does Python's json module do with duplicate keys? How did you merge them?

Python's json.loads() does not preserve duplicate keys. When duplicate keys are present, the last value normally replaces the earlier value. In this example, "a":1 is replaced by "a":3.

To merge duplicate keys, I used object_paris_hook=list so that all key-value pairs are preserved. Then I looped through the paris and added to values for matching keys. Therefore, the final result is {'a': 4, 'b': 2}.

Question 5: Fixing JSON with Comments and Trailing Commas

The following string is not valid JSON because it contains // comments and a trailing comma. Remove the comments and the trailing comma, then parse it into a Python dictionary and print it.


```
{
  // user name
  "name": "Alice",
  "age": 30, // age in years
  "city": "New York",
}
```



````python
import json
import re

invalid_json = '''
{
  // user name
  "name": "Alice",
  "age": 30, // age in years
  "city": "New York",
}
'''

cleaned_json = re.sub(r'//.*', '', invalid_json)
cleaned_json = re.sub(r',\s*\}', '}', cleaned_json)

data = json.loads(cleaned_json)
print(data)
````

Explanation: Describe the steps you took to clean the string before parsing.

I removed the // comments using a regular expression and removed trailing commas before closing braces or brackets. Then I used json.loads() to convert the cleaned JSON string into a Python dictionary.

Question 6: Validating and Formatting a JSON String

Write a function validate_and_format(json_string) that:
- Attempts to parse the given JSON string.
- If parsing fails, returns "Invalid JSON".
- If parsing succeeds, returns the JSON string pretty-printed with indentation of 4 spaces and sorted keys.
- Test it with both a valid and an invalid JSON string.

````python
import json

def validate_and_format(json_string):
    try:
        data = json.loads(json_string)
        return json.dumps(data, indent=4, sort_keys=True)
    except json.JSONDecodeError as e:
        return f"Invalid JSON: {e}"

# Test cases
valid = '{"b": 2, "a": 1}'
invalid = '{"a": 1, "b": 2,}'

print(validate_and_format(valid))
print(validate_and_format(invalid))
````
